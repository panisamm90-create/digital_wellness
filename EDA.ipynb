```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OrdinalEncoder, OneHotEncoder
from sklearn.pipeline import Pipeline
import xgboost as xgb
```

```python
df = pd.read_csv('dataset/digital_wellness_dataset.csv')
df.head()
```

```python
df['date'] = pd.to_datetime(df['date'])
df['date'] = df['date'].dt.day

df.head()
```

```python
cat_cols = df.select_dtypes(include=['object', 'category', 'bool']).columns.to_list()
print(cat_cols)
num_cols = df.select_dtypes(include=['int64', 'float64']).columns.to_list()
print(num_cols)
```

```python
print(df[num_cols].mean())

df.groupby('unhealthy_day')['total_screen_minutes'].mean().plot(kind='bar')
```

```python
age_order = ['13-15', '16-18', '19-24', '25-34', '35-50']
ordinal = OrdinalEncoder(categories=[age_order])
df['age_group_encoded'] = ordinal.fit_transform(df[['age_group']])
```

```python
df = df.sort_values(['date'])
```

```python
X = df.drop('unhealthy_day', axis=1)
Y = df.unhealthy_day

X_Train, X_Test, Y_Train, Y_Test = train_test_split(X, Y, test_size=0.2,stratify=Y)
```

```python
cat_cols = X.select_dtypes(include=['object', 'category', 'bool']).columns.to_list()
num_cols = X.select_dtypes(include=['int64', 'float64']).columns.to_list()

print(num_cols)
```

```python
preproccessor = ColumnTransformer([
    ('num', StandardScaler(), num_cols),
    ('cat', OneHotEncoder(handle_unknown='ignore'), cat_cols)],
    remainder='passthrough')

preproccessor
```

```python
pip = Pipeline([
    ('preprocess', preproccessor),
    ('model', RandomForestClassifier(n_estimators=200, max_depth=9)),
])
```

```python
pip.fit(X_Train, Y_Train)
```

```python
predict_class = pip.predict(X_Test)


from sklearn.metrics import classification_report

print(classification_report(Y_Test, predict_class))
```

```python
"""
Digital Wellness Recommendation System
این ماژول دو کار مربوط به ردیف‌های ۴ و ۵ جدول را پیاده می‌کند:

  Task 4 - Generate wellness suggestions (Rule-Based, NOT ML)
      ورودی: خروجی مدل کلاسیفیکیشن (healthy/unhealthy + confidence)
             + خروجی مدل رگرشن (پیش‌بینی screen time فردا)
             + مقادیر خام فیچرها (raw feature values)
      خروجی: لیستی از پیشنهادها (List[str])

  Task 5 - Personalize recommendations over time (Rule-Based, NOT ML)
      ورودی: rolling averages کاربر، شغل (job)، اپ‌های استفاده‌شده، مقدار امروز
      خروجی: True/False برای هر شاخص (آیا نسبت به میانگین گذشته «انحراف»
             دارد) + پیام شخصی‌سازی‌شده متناسب با آن

هیچ کدام از این دو تابع مدل ML جدیدی train نمی‌کنند؛ فقط از خروجی
مدل‌های Task 2 (classification) و Task 3 (regression) که در نوت‌بوک
EDA ساخته شده‌اید به عنوان ورودی استفاده می‌کنند و روی آن‌ها قانون
(if/else) اعمال می‌کنند.
"""

from dataclasses import dataclass, field
from typing import Dict, List, Optional



# پیشنهادهای سلامت


def generate_wellness_suggestions(
    classification_label: str,          # healthy or unhealthy
    classification_confidence: float,   # model confidence
    predicted_tomorrow_minutes: float,  # tomorrow prediction
    today_total_minutes: float,         # today's total
    raw_features: Dict[str, float],     # raw values for the day
) -> List[str]:
    """
    قوانین ثابت (بدون یادگیری ماشین) برای تولید لیست پیشنهادها.
    raw_features می‌تواند شامل کلیدهایی مثل موارد زیر باشد:
        late_night_usage (ساعت), sleep_hours (ساعت),
        social_media_minutes, work_related_minutes,
        num_unlocks, screen_time_before_bed_minutes ...
    فقط کلیدهایی که وجود دارند بررسی می‌شوند، پس نیازی نیست
    همه فیچرها را پاس بدهید.
    """
    suggestions: List[str] = []

    # نتیجه‌ی مدل را به یک پیشنهاد تبدیل می‌کنیم.
    if classification_label == "unhealthy":
        if classification_confidence >= 0.75:
            suggestions.append(
                "الگوی امروز شما با اطمینان بالا «ناسالم» تشخیص داده شد؛ "
                "بهتر است همین امروز محدودیت روزانه اپ‌ها را فعال کنید."
            )
        else:
            suggestions.append(
                "نشانه‌هایی از استفاده‌ی ناسالم دیده می‌شود (اطمینان مدل متوسط است)؛ "
                "وضعیت را یکی‌دو روز دیگر هم زیر نظر داشته باشید."
            )
    else:
        suggestions.append("الگوی امروز شما در محدوده‌ی سالم قرار دارد. همین روند را ادامه دهید.")

    # دلیل احتمالی وضعیت را از روی داده‌های خام بررسی می‌کنیم.
    late_night = raw_features.get("late_night_usage")
    if late_night is not None and late_night > 1:
        suggestions.append(f"استفاده‌ی شبانه شما {late_night} ساعت بوده؛ استفاده از موبایل بعد از ساعت ۱۱ شب را کم کنید.")

    sleep_hours = raw_features.get("sleep_hours")
    if sleep_hours is not None and sleep_hours < 6:
        suggestions.append(f"ساعت خواب شما ({sleep_hours} ساعت) کمتر از حد توصیه‌شده است؛ سعی کنید حداقل ۷ ساعت بخوابید.")

    social_media = raw_features.get("social_media_minutes")
    if social_media is not None and social_media > 120:
        suggestions.append("زمان شبکه‌های اجتماعی امروز بالا بوده؛ برای این دسته از اپ‌ها تایمر یادآوری تنظیم کنید.")

    num_unlocks = raw_features.get("num_unlocks")
    if num_unlocks is not None and num_unlocks > 80:
        suggestions.append(f"تعداد باز کردن گوشی امروز ({num_unlocks} بار) زیاد است؛ نوتیفیکیشن‌های غیرضروری را خاموش کنید.")

    # پیش‌بینی فردا را هم در نظر می‌گیریم.
    if predicted_tomorrow_minutes > today_total_minutes * 1.1:
        suggestions.append(
            f"مدل پیش‌بینی می‌کند فردا حدود {predicted_tomorrow_minutes:.0f} دقیقه "
            f"استفاده کنید (بیشتر از امروز)؛ از همین امروز یک هدف روزانه برای فردا تعیین کنید."
        )
    elif predicted_tomorrow_minutes < today_total_minutes * 0.9:
        suggestions.append("پیش‌بینی می‌شود فردا کمتر از امروز از گوشی استفاده کنید؛ روند رو به بهبود است.")

    if not suggestions:
        suggestions.append("در حال حاضر پیشنهاد خاصی نیاز نیست، به همین روند ادامه دهید.")

    return suggestions



# شخصی‌سازی پیشنهادها


@dataclass
class PersonalizationResult:
    is_spike: bool                       # whether the value is a spike
    metric: str                          # metric name
    today_value: float
    rolling_average: float
    ratio: float
    message: Optional[str] = None


def personalize_over_time(
    job: str,
    used_apps: List[str],
    today_values: Dict[str, float],      # today's values
    rolling_averages: Dict[str, float],  # recent averages
    spike_threshold: float = 1.3,        # spike cutoff
) -> List[PersonalizationResult]:
    """
    برای هر شاخصی که هم در today_values و هم در rolling_averages وجود دارد،
    بررسی می‌کند آیا امروز نسبت به میانگین تاریخی کاربر «جهش» (spike) داشته.
    خروجی هر شاخص یک True/False است (دقیقا مطابق ستون Output ردیف ۵ جدول)
    به‌همراه پیام شخصی‌سازی‌شده متناسب با شغل (job) و اپ‌های استفاده‌شده.
    """
    results: List[PersonalizationResult] = []

    for metric, today_value in today_values.items():
        avg = rolling_averages.get(metric)
        if avg is None or avg == 0:
            continue

        ratio = today_value / avg
        is_spike = ratio > spike_threshold

        message = None
        if is_spike:
            message = _build_personalized_message(job, used_apps, metric, ratio)

        results.append(
            PersonalizationResult(
                is_spike=is_spike,
                metric=metric,
                today_value=today_value,
                rolling_average=avg,
                ratio=ratio,
                message=message,
            )
        )

    return results


def _build_personalized_message(job: str, used_apps: List[str], metric: str, ratio: float) -> str:
    """پیام را بر اساس شغل کاربر و اپ‌های پراستفاده تنظیم می‌کند."""
    job = (job or "").lower()

    base = f"«{metric}» امروز {ratio:.1f} برابر میانگین معمول شما شده."

    if "student" in job or "دانش" in job:
        tip = "اگر این افزایش به‌خاطر اپ‌های سرگرمی بوده، بهتر است قبل از امتحان‌ها محدودش کنید."
    elif "remote" in job or "دورکار" in job:
        tip = "اگر بخشی از این افزایش کاری بوده طبیعی است، اما استراحت‌های کوتاه بین کار را فراموش نکنید."
    else:
        tip = "سعی کنید فردا این شاخص را به سطح میانگین معمول خودتان برگردانید."

    if used_apps:
        top_apps = "، ".join(used_apps[:3])
        tip += f" (اپ‌های پراستفاده‌ی امروز: {top_apps})"

    return base + " " + tip



# ترکیب دو بخش پیشنهادها


def get_full_recommendation(
    classification_label: str,
    classification_confidence: float,
    predicted_tomorrow_minutes: float,
    today_total_minutes: float,
    raw_features: Dict[str, float],
    job: str,
    used_apps: List[str],
    today_values: Dict[str, float],
    rolling_averages: Dict[str, float],
) -> Dict[str, object]:
    """یک خروجی یکپارچه شامل هم پیشنهادهای عمومی (Task 4) و هم
    نتایج شخصی‌سازی‌شده‌ی زمانی (Task 5) برمی‌گرداند."""

    general_suggestions = generate_wellness_suggestions(
        classification_label,
        classification_confidence,
        predicted_tomorrow_minutes,
        today_total_minutes,
        raw_features,
    )

    personalization = personalize_over_time(job, used_apps, today_values, rolling_averages)

    return {
        "general_suggestions": general_suggestions,
        "personalization": personalization,
        "personalized_messages": [p.message for p in personalization if p.message],
    }



# یک نمونه برای تست تابع

if __name__ == "__main__":
    # این مقادیر در استفاده‌ی واقعی از خروجی مدل‌ها می‌آیند.

    example_classification_label = "unhealthy"
    example_classification_confidence = 0.82
    example_predicted_tomorrow = 360.0

    example_raw_features = {
        "late_night_usage": 1.5,
        "sleep_hours": 5.5,
        "social_media_minutes": 150,
        "num_unlocks": 95,
    }

    result = get_full_recommendation(
        classification_label=example_classification_label,
        classification_confidence=example_classification_confidence,
        predicted_tomorrow_minutes=example_predicted_tomorrow,
        today_total_minutes=300,
        raw_features=example_raw_features,
        job="student",
        used_apps=["Instagram", "YouTube", "TikTok"],
        today_values={"total_screen_minutes": 300, "social_media_minutes": 150},
        rolling_averages={"total_screen_minutes": 210, "social_media_minutes": 90},
    )

    print("=== General Suggestions (Task 4) ===")
    for s in result["general_suggestions"]:
        print("-", s)

    print("\n=== Personalization (Task 5) ===")
    for p in result["personalization"]:
        print(p)

    print("\n=== Personalized Messages ===")
    for m in result["personalized_messages"]:
        print("-", m)
```
